```python
import pandas as pd
import numpy as np
import re
```

```python
df = pd.read_excel('all_reviews.xlsx')
```

```python
df
```

```python
df['Адрес точки'] = df['Адрес точки'].replace('Не указан', np.nan)
df['Оценка'] = df['Оценка'].replace('Форум (без оценки)', np.nan)
df['Оценка'] = pd.to_numeric(df['Оценка'])
```

```python
df['Город'] = df['Адрес точки'].str.split(',').str[0]
```

```python
df
```

```python
months = {
    'января': '01',
    'февраля': '02',
    'марта': '03',
    'апреля': '04',
    'мая': '05',
    'июня': '06',
    'июля': '07',
    'августа': '08',
    'сентября': '09',
    'октября': '10',
    'ноября': '11',
    'декабря': '12',

    'янв': '01',
    'фев': '02',
    'мар': '03',
    'апр': '04',
    'май': '05',
    'июн': '06',
    'июл': '07',
    'авг': '08',
    'сен': '09',
    'сент': '09',
    'окт': '10',
    'ноя': '11',
    'дек': '12'
}

def replace_dates(date):
    '''
    замена месяцев на цифры в столбце 
    '''
    date = str(date).lower()

    for month, numer in months.items():
        date = re.sub(rf'\b{month}\b', numer, date)

    return date

def add_year(date):
    date = str(date).lower().strip()

    if len(date.split()) == 2:
        date = date + ' 2026'

    return date

def drop_time(date):
    date = str(date).lower().strip()

    if ' в ' in date:
        date = date.split(' в ')[0]
        
    return date 

df['Дата'] = df['Дата'].apply(replace_dates)
df['Дата'] = df['Дата'].apply(add_year)
df['Дата'] = df['Дата'].apply(drop_time)
```

```python
df['Дата'] = pd.to_datetime(df['Дата'], format='mixed', dayfirst=True)
```

```python
df
```

```python
df.to_excel('reviews_processed.xlsx', index=False)
```
